```python
class Stack:
    def __init__(self):
        self.items = []
        self.size = 0
        
    def push(self, item):
        self.items.append(item)
        self.size += 1
        return self.items

    def pop(self):
        if self.size == 0:
            return None
        item = self.items[self.size - 1]
        self.items = self.items[:self.size - 1]
        self.size -= 1
        return item

    def is_empty(self):
        return self.size == 0

    def get_size(self):
        return self.size

    def display(self):
        for item in self.items:
            print(item, end="->")
        print()
```

```python
s = Stack()
print(s.is_empty())
s.push("A")
s.push("B")
s.pop()
s.push("C")
s.push("D")
s.push("E")
s.pop()
s.pop()
s.push("F")
s.push("G")
s.display()
print(s.is_empty())
print(s.get_size())
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.top = None
        self.count = 0

    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        self.count += 1

    def pop(self):
        if self.is_empty():
            raise IndexError("Stack is Empty")
    
        data = self.top.data
        self.top = self.top.next
        self.count -= 1
        return data

    def is_empty(self):
        return self.top is None

    def size(self):
        return self.count

    def display(self):
        current = self.top

        while current is not None:
            print(current.data)
            current = current.next
```

```python
s = Stack()
print(s.is_empty())
s.push("A")
s.push("B")
s.pop()
s.push("C")
s.push("D")
s.push("E")
s.pop()
s.pop()
s.push("F")
s.push("G")
s.display()
print(s.is_empty())
print(s.size())
```

```python
class Queue:
    def __init__(self):
        self.items = []
        self.count = 0
        
    def enqueue(self, item):
        self.items.append(item)
        self.count += 1
        return self.items
    
    def dequeue(self):
        if self.is_empty():
            raise IndexError("Stack is Empty")
        item = self.items[0]
        self.items = self.items[1:]
        self.count -= 1
        return item

    def is_empty(self):
        return self.count == 0

    def size(self):
        return self.count

    def display(self):
        for item in self.items:
            print(item, end="->")
        print()
```
